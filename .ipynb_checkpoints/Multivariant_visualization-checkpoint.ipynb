```python
# Multivariate Data Visualization
### Scatter Plot Matrix, Heatmap, Bubble Plot, Parallel Coordinate Plot
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
df.head()
```

```python
df.info()
```

```python
sns.pairplot(df, hue='species')
plt.show()
```

```python
pd.plotting.scatter_matrix(
    df.iloc[:, :4],
    figsize=(10, 10)
)
plt.suptitle("Scatter Plot Matrix - Iris Dataset", y=1.02)
plt.show()
```

```python
corr = df.iloc[:, :4].corr()
corr
```

```python
# Remove 'species' column because correlation works only on numerical data
corr = df.drop('species', axis=1).corr()

# Create a new figure with size 8x6 inches (makes plot bigger and readable)
plt.figure(figsize=(8,6))

# Draw heatmap
sns.heatmap(
    corr,              # Correlation matrix data
    annot=True,        # Write correlation values inside boxes
    cmap='coolwarm'    # Color scheme (red = positive, blue = negative)
)

# Display the plot
plt.show()
```

```python
# Show correlation matrix as an image
plt.imshow(corr)

# Add a color bar on the side to show color scale
plt.colorbar()

# Add feature names on x-axis
plt.xticks(
    range(len(corr.columns)),  # Position of ticks
    corr.columns,              # Column names as labels
    rotation=45                # Rotate for better readability
)

# Add feature names on y-axis
plt.yticks(
    range(len(corr.columns)),  # Position of ticks
    corr.columns               # Column names as labels
)

# Add title to the plot
plt.title("Correlation Heatmap")

# Display the plot
plt.show()
```

```python
plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    s=df['petal length (cm)'] * 20
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Bubble Plot (Bubble Size = Petal Length)")
plt.show()

```

```python
df_parallel = df.copy()
df_parallel['species'] = df_parallel['species'].astype(str)
df_parallel.head()
```

```python
from pandas.plotting import parallel_coordinates

parallel_coordinates(
    df_parallel[['sepal length (cm)', 'sepal width (cm)',
                 'petal length (cm)', 'petal width (cm)', 'species']],
    'species'
)

plt.title("Parallel Coordinate Plot - Iris Dataset")
plt.show()
```
