```python
jan = pd.DataFrame({
    "ProductID": [1, 2, 1],
    "Sales": [200, 150, 100]
})

feb = pd.DataFrame({
    "ProductID": [1, 2, 2],
    "Sales": [250, 100, 200]
})
```

```python
jan
```

```python
feb
```

```python
all_sales = pd.concat([jan, feb], keys=["Jan", "Feb"], names=["Month"])
all_sales = all_sales.reset_index(level=0)
all_sales
```

```python
products = pd.DataFrame({
    "ProductID": [1, 2],
    "ProductName": ["Laptop", "Phone"]
})

merged = all_sales.merge(products, on="ProductID")
merged
```

```python
products.to_csv("product.csv")
```

```python
pivot = pd.pivot_table(
    merged,
    index="ProductName",
    columns="Month",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)
pivot
```

```python
ct = pd.crosstab(merged["ProductName"], merged["Month"])
ct
```

```python
percentage = pivot.div(pivot.sum().sum()) * 100
percentage
```

```python
df1 = pd.DataFrame({
    "ProductID": [101, 102, 103, 101],
    "Region": ["North", "South", "East", "West"],
    "Sales": [12000, 15000, 10000, 9000]
})

df1
```

```python
df2 = pd.DataFrame({
    "ProductID": [101, 102, 103, 104],
    "Region": ["North", "South", "East", "West"],
    "Sales": [14000, 16000, 11000, 13000]
})

df2
```

```python
df3 = pd.DataFrame({
    "ProductID": [101, 102, 104, 103],
    "Region": ["North", "South", "East", "West"],
    "Sales": [15500, 17000, 12500, 9500]
})

df3
```

```python
df=pd.concat([df1,df2,df3])
df
```

```python
price_df = pd.DataFrame({
    "ProductID": [101, 102, 103, 104],
    "Price": [50, 40, 30, 60]
})

price_df
```

```python
merged_df = df.merge(price_df, on="ProductID", how="inner")
merged_df
```

```python
merged_df["Revenue"] = np.multiply(
    merged_df["Sales"].values,
    merged_df["Price"].values
)

merged_df
```

```python
revenue_per_product = merged_df.groupby("ProductID")["Revenue"].sum()

revenue_per_product
```

```python
survey = pd.DataFrame({
    "Department": ["CS", "CS", "IT", "IT", "CS", "ECE"],
    "Grade": ["A", "B", "A", "B", "A", "A"]
})

survey
```

```python
ct = pd.crosstab(survey["Department"], survey["Grade"])
ct
```

```python
percentage_ct = np.round(
    (ct / ct.sum(axis=1).values.reshape(-1, 1)) * 100,
    2
)

percentage_ct
```
