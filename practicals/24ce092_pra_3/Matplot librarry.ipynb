```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
```

```python
year=[1972,1982,1992,2002,2012]
e_india=[100.6,158.61,305.54,394.96,724.79]
e_bangladesh=[10.5,25.21,58.65,119.27,274.87]

fig=plt.figure(figsize=(8,6))
```

```python
plt.plot(year,e_india,color='orange', label='india', marker='*', markersize=12)
plt.plot(year, e_bangladesh, color='g' , label='Bangladesh', linestyle='dashed')

font1={'family':'serif','color':'blue','size':20}
font2={'family':'serif','color':'green', 'size':15}
plt.grid()

plt.xlabel('year', fontdict=font1)
plt.ylabel(' power consuption in kwh', fontdict=font2)

plt.title('Electricity consuption per capita' , fontdict= font2)

pop_a= mpatches.Patch(color = '#5DC83F', label='india')
pop_b= mpatches.Patch(color='#CE5D45' , label='bangladesh')

plt.legend(handles= [pop_a,pop_b],frameon=True, labelspacing=2, ncol=2)
plt.savefig('electricity_plot.jpg', bbox_inches='tight', dpi=100, format='jpeg')
plt.show()
```

```python
plt.plot(year,e_india,color='orange', label='india', marker='*', markersize=12)
plt.plot(year, e_bangladesh, color='g' , label='Bangladesh', linestyle='dashed')

font1={'family':'serif','color':'blue','size':20}
font2={'family':'serif','color':'green', 'size':15}
plt.grid()

plt.xlabel('year', fontdict=font1)
plt.ylabel(' power consuption in kwh', fontdict=font2)

plt.title('Electricity consuption per capita of india and bangladesh' , fontdict= font2)

pop_a= mpatches.Patch(color = '#5DC83F', label='india')
pop_b= mpatches.Patch(color='#CE5D45' , label='bangladesh')

plt.legend(handles= [pop_a,pop_b],frameon=True, labelspacing=10, ncol=10)
plt.show()
```

## subplot

```python
#plot 1
x=np.array([0,1,2,2,3])
y=np.array([3,8,1,7,10])

plt.subplot(2,2,1) # 1 has row, 2 has columns, and this plot has first figure

plt.title("PLOT : 1")
plt.plot(x,y)

#plot 2
x=np.array([0,1,2,3])
y=np.array([10,20,30,40])

plt.subplot(2,2,2)

plt.title("PLOT : 2")
plt.plot(x,y)
plt.subplots_adjust(hspace=0.4)

#plot 3
x=np.array([0,1,4,3])
y=np.array([20,50,30,40])

plt.subplot(2,2,3)

plt.title("PLOT : 3")
plt.plot(x,y)

plt.subplots_adjust(hspace=0.4)

plt.savefig('tight_plot.jpg', bbox_inches='tight', dpi=100, format='jpeg')

plt.show()
```
