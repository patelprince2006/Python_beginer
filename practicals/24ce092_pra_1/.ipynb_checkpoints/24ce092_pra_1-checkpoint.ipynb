```python
indata={
   'roll_no' : [1,2,3,4,5],
    'name' : ["prince","a","b","c","d"],
    'chemestry': [95,84,86,79,91],
    'physics':[92,86,75,68,83],
    'maths':[78,85,95,84,76],
}
df=pd.DataFrame(indata)
df
```

```python
df.set_index(['roll_no'])
```

```python
print("using pandas",df[['chemestry', 'physics', 'maths']].describe())
```

```python
marks= [95,84,86,79,91]
print("------using numpy------")
print("Max Marks", np.max(marks))
print("Min Marks",np.min(marks))
print("Mean Marks",np.mean(marks))
print("Std Dev",np.std(marks))
```

```python
marks_cols=["chemestry","physics","maths"]
df['Percentage'] = (df[marks_cols].sum(axis=1).to_numpy() / 300) * 100
df
```

```python
numeric_df = df.select_dtypes(include=np.number)

data = numeric_df.values

mean_values = np.mean(data, axis=0)
median_values = np.median(data, axis=0)

print("\nColumn-wise Mean:", mean_values)
print("Column-wise Median:", median_values)

row_totals = np.sum(data[:, :3], axis=1) 
max_index = np.argmax(row_totals)

print("\nStudent with highest total marks:")
print(df.iloc[max_index])
```
