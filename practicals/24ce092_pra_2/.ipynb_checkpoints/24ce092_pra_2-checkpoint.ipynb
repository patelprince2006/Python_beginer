```python
df=pd.DataFrame({
    'Student_ID': [1, 2, 3, 4, 5],
    'Name': ['A', 'B', 'C', 'E', 'D'],
    'Gender': ['M', 'Female', 'male', 'F', 'Male'],
    'Marks': [85, np.nan, 78, 90, None],
    'City': ['Delhi', 'Mumbai', 'Mumbai', 'Delhi', 'mumbai']
})
```

```python
print("before : ",df)

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Gender'] = df['Gender'].replace(
    {'M':'Male', 'male':'Male', 'F':'Female'}
)
print("after : ",df)
```

```python
g=df.groupby('City')
g
```

```python
agg=df.groupby('City')['Marks'].agg(
        mean='mean',
        max='max',
        count='count'
)
agg
```
