## Missing value Analysis

```python
full_clean = full.copy()
```

```python
full_clean = full.copy()
for col in ["Gender","Embarked"] :
 if col in full_clean.columns:
  full_clean[col]=full_clean[col].astype(str).str.strip()
  full_clean.loc[full_clean[col].isin(["nan","NONE","NaN",""]),col]=np.nan #iloc

full_clean[["Gender","Embarked"]].head(10)
```

```python
if "cabin" in full_clean.columns:
     full_clean.drop(columns=["cabin"],inplace=True)

if "Embarked" in full_clean.columns:
    full_clean["Embarked"] = full_clean["Embarked"].fillna(full_clean["Embarked"].mode(dropna=True)[0])
```

```python
if "Age" in full_clean.columns:
    full_clean["Age"]=full_clean["Age"].fillna(full_clean["Age"].median())
```

```python
if  "Fare" in full_clean.columns:
    full_clean["Fare"]=full_clean["Fare"].fillna(full_clean["Fare"].median())
```
